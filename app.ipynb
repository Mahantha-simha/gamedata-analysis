```python
import pandas as pd
import numpy as np

pd.set_option('future.no_silent_downcasting', True)
pd.set_option('display.max_columns', None)


df1 = pd.read_json('GameFactMockData2.json')
df2=pd.json_normalize(df1['gamefact_data'])

df=df2.copy()
```

```python
df.replace('None', np.nan, inplace=True)
df.replace('Invalid', np.nan, inplace=True)
```

```python
dro = df[df[['EventID', 'StadiumID', 'RefereeID', 'TimeID', 'DateID']].isna().any(axis=1)].index
df.drop(dro, inplace=True)
```
